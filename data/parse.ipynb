```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
sv_chats = pd.read_csv("sv_chats.csv")
```

```python
chat_idx_to_id = dict()
for i,row in sv_chats.iterrows():
    chat_idx_to_id[row['chat_id']] = i
```

```python
sv = pd.read_parquet("sv_to_sv.parquet")
```

```python
sv.head(10)
```

```python
plt.figure(figsize=((10,3)))
plt.hist(sv['date'],bins=60)
plt.title("Forwards in Svedish Chats over Time")
```

```python
from_ = [chat_idx_to_id[i] for i in sv['from_chat_id']]
```

```python
to_ = [chat_idx_to_id[i] for i in sv['to_chat_id']]
```

```python
df = pd.DataFrame({"msg_id":sv['msg_id'],"from":from_,"to":to_,"date":sv['date']})
```

```python
df.to_csv("sv_forwards.csv",index=False)
```
